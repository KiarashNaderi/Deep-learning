```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
```

**Data definition**

```python
x=tf.constant([-3,-5,0,1,4,9,10],dtype=float)
y=x*3-3
```

```python
x.numpy()
```

```python
y.numpy()
```

**Model definition**

```python
model=keras.Sequential([
    keras.layers.Dense(units=1,activation=None,input_shape=[1])
])
# sequential  yek model piade siazi shabake asabi
#dense==>  yani tak take neuron haye yek laye be nueoron haye layey dg vasland
#units==> tedad neuron ha
# inupt shapes==> yani faghat yek vrodi mikhaim begirim(tedad feature)
```

**Model compile**

```python
model.compile(optimizer="sgd",loss="mean_squared_error")
# optimizer==> noe behine sazeton
#loss==> noe tabe losseton==> baraye mizane bad budane model
```

**model summary**

```python
model.summary()
# dar keras niazi be tanzime input nist
# 1 vorodi darim be layeye dense==> 2 param==> har neuron 1 vrodi darad va har vrodi yek vazn darad va kole yek laye yek bayas darad==>
# 1 vrodi= 1 vazn + 1 bayas=2 paramtre daryeye dense 1
```

**Model Training**

```python
model.fit(x,y,batch_size=1,epochs=500)
#batch size==> pack bandi data
#epochs=>>kole faraynd chand bar tey shavad
```

**test kardan**

```python
x_test=tf.constant([-2,7,14],dtype=float)
y_test=x_test*3-3
```

```python
x_test.numpy()
```

```python
y_test.numpy()
```

```python
y_predict=model.predict(x_test)
```
