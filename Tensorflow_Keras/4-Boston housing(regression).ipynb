```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import boston_housing

# بارگذاری دیتاست Boston Housing
(train_data, train_targets), (test_data, test_targets) = boston_housing.load_data()

# نرمال‌سازی داده‌ها: هر ویژگی با کم کردن میانگین و تقسیم بر انحراف معیار نرمال می‌شود
mean = train_data.mean(axis=0)
std = train_data.std(axis=0)
train_data = (train_data - mean) / std
test_data = (test_data - mean) / std

# ساخت مدل شبکه عصبی
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(train_data.shape[1],)),  # لایه Dense با 64 نرون و تابع فعال‌سازی ReLU
    layers.Dense(64, activation='relu'),  # لایه Dense دیگر با 64 نرون و تابع فعال‌سازی ReLU
    layers.Dense(1)  # لایه خروجی برای پیش‌بینی یک مقدار عددی (قیمت خانه)
])

# کامپایل مدل: استفاده از RMSprop به‌عنوان optimizer و mse به‌عنوان loss function
model.compile(optimizer='rmsprop',
              loss='mse',
              metrics=['mae'])

# آموزش مدل
history = model.fit(train_data, train_targets, epochs=100, batch_size=10, validation_split=0.2)

# ارزیابی مدل بر روی داده‌های تست
test_mse_score, test_mae_score = model.evaluate(test_data, test_targets)
print(f'Test MAE: {test_mae_score}')
```
